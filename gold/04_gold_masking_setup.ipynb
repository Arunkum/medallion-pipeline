```python
%sql
CREATE SCHEMA IF NOT EXISTS healthcare_dev.governance;

CREATE OR REPLACE FUNCTION healthcare_dev.governance.mask_string(col_value STRING)
RETURNS STRING
RETURN
  CASE
    WHEN is_account_group_member('phi_authorized_users') THEN col_value
    ELSE 'REDACTED'
  END;

CREATE OR REPLACE FUNCTION healthcare_dev.governance.mask_date(col_value DATE)
RETURNS DATE
RETURN
  CASE
    WHEN is_account_group_member('phi_authorized_users') THEN col_value
    ELSE NULL
  END;
```

```python
%sql
ALTER TABLE healthcare_dev.gold.csv_admissions
ALTER COLUMN mrn SET MASK healthcare_dev.governance.mask_string;

ALTER TABLE healthcare_dev.gold.csv_admissions
ALTER COLUMN patient_first_name SET MASK healthcare_dev.governance.mask_string;

ALTER TABLE healthcare_dev.gold.csv_admissions
ALTER COLUMN patient_last_name SET MASK healthcare_dev.governance.mask_string;

ALTER TABLE healthcare_dev.gold.csv_admissions
ALTER COLUMN date_of_birth SET MASK healthcare_dev.governance.mask_date;
```
